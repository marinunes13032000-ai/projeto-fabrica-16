## ETAPA 7 · Visualização

**Regra:** cada gráfico responde a 1 pergunta. Título e eixos nomeados, sempre.

## 11.1. Gráfico 1

> ✍️ **Sua missão:** Qual plataforma concentra mais horas de jogo? — dica: `sns.barplot(data=df, x='plataforma', y='horas_jogadas')`  
> Depois, escreva **1 frase** interpretando o gráfico.

```python
sns.barplot(data=df, x='plataforma', y='horas_jogadas')
plt.title('Horas de jogo por plataforma')
plt.ylabel('Horas')
plt.show()
```

## 11.2. Gráfico 2

> ✍️ **Sua missão:** Existe relação entre horas jogadas e pontuação? — dica: `sns.scatterplot`  
> Depois, escreva **1 frase** interpretando o gráfico.

```python
sns.scatterplot(data=df, x='horas_jogadas', y='pontuacao')
plt.title('Horas jogadas x Pontuação')
plt.show()
```

### Tem uma relação, pois no gráfico mostra a quantidade de horas que foi jogado e as pontuações. É de se analisar que com menos tempo jogando tem mais pontuações e com mais tempo jogando consiste em ter menos pontuação.

## 11.3. Gráfico 3

```python
sns.boxplot(data=df, x='jogo', y='pontuacao')
plt.title('Distribuição da pontuação por jogo')
plt.show()
```
